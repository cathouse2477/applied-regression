第 29 條
從價課徵關稅之進口貨物，其完稅價格以該進口貨物之交易價格作為計算根據。<br>
前項交易價格，指進口貨物由輸出國銷售至中華民國實付或應付之價格。<br>
進口貨物之實付或應付價格，如未計入下列費用者，應將其計入完稅價格：<br>
一、由買方負擔之佣金、手續費、容器及包裝費用。<br>
二、由買方無償或減價提供賣方用於生產或銷售該貨之下列物品及勞務，<br>
    經合理攤計之金額或減價金額：<br>
 (一) 組成該進口貨物之原材料、零組件及其類似品。<br>
 (二) 生產該進口貨物所需之工具、鑄模、模型及其類似品。<br>
 (三) 生產該進口貨物所消耗之材料。<br>
 (四) 生產該進口貨物在國外之工程、開發、工藝、設計及其類似勞務。<br>
三、依交易條件由買方支付之權利金及報酬。<br>
四、買方使用或處分進口貨物，實付或應付賣方之金額。<br>
五、運至輸入口岸之運費、裝卸費及搬運費。<br>
六、保險費。<br>
依前項規定應計入完稅價格者，應根據客觀及可計量之資料。無客觀及可計量之資料者，視為無法按本條規定核估其完稅價格。<br>
海關對納稅義務人提出之交易文件或其內容之真實性或正確性存疑，納稅義務人未提出說明或提出說明後，海關仍有合理懷疑者，視為無法<br>按本條規定核估其完稅價格。<br>

第 31 條<br>
進口貨物之完稅價格，未能依第二十九條規定核定者，海關得按該貨物出口時或出口前、後銷售至中華民國之同樣貨物之交易價格核定之。核定時，應就交易型態、數量及運費等影響價格之因素作合理調整。<br>
前項所稱同樣貨物，指其生產國別、物理特性、品質及商譽等均與該進口貨物相同者。<br>

第 32 條<br>
進口貨物之完稅價格，未能依第二十九條及前條規定核定者，海關得按該貨物出口時或出口前、後銷售至中華民國之類似貨物之交易價格核定之。核定時，應就交易型態、數量及運費等影響價格之因素作合理調整。<br>
前項所稱類似貨物，指與該進口貨物雖非完全相同，但其生產國別及功能相同，特性及組成之原材料相似，且在交易上可互為替代者。

第 33 條<br>
進口貨物之完稅價格，未能依第二十九條、第三十一條及前條規定核定者，海關得按國內銷售價格核定之。<br>
海關得依納稅義務人請求，變更本條及第三十四條核估之適用順序。<br>
第一項所稱國內銷售價格，指該進口貨物、同樣或類似貨物，於該進口貨物進口時或進口前、後，在國內按其輸入原狀於第一手交易階段，售予無特殊關係者最大銷售數量之單位價格核計後，扣減下列費用計算者：<br>
一、該進口貨物、同級或同類別進口貨物在國內銷售之一般利潤、費用或<br>
    通常支付之佣金。<br>
二、貨物進口繳納之關稅及其他稅捐。<br>
三、貨物進口後所發生之運費、保險費及其相關費用。<br>

第 34 條<br>
進口貨物之完稅價格，未能依第二十九條、第三十一條、第三十二條及前條規定核定者，海關得按計算價格核定之。<br>
前項所稱計算價格，指下列各項費用之總和：<br>
一、生產該進口貨物之成本及費用。<br>
二、由輸出國生產銷售至中華民國該進口貨物、同級或同類別貨物之正常<br>
    利潤與一般費用。<br>
三、運至輸入口岸之運費、裝卸費、搬運費及保險費。<br>

第 35 條<br>
進口貨物之完稅價格，未能依第二十九條、第三十一條、第三十二條、第三十三條及前條規定核定者，海關得依據查得之資料，以合理方法核定之。<br>
